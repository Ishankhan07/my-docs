# error detection

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12,8))
sns.boxplot(dfs)
```

```python
dfs.columns
```

```python
for i in range(60):
    q1=dfs[i].quantile(0.25)
    q3=dfs[i].quantile(0.75)
    iqr = q3-q1
    lower = q1-(1.5*iqr)
    upper = q3+(1.5*iqr)
    dfs[i] = dfs[i].map(lambda x : lower if x<lower else upper if x>upper else x)
```

```python
plt.figure(figsize=(12,8))
sns.boxplot(dfs)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
dfs[60]=le.fit_transform(dfs[60])
dfs[60].unique()
```

```python
x = dfs.drop(60,axis=1)
y = dfs[60]
```

```python
x
```

```python
from imblearn.over_sampling import RandomOverSampler
ros = RandomOverSampler()
```

```python
xros,yros = ros.fit_resample(x,y)
```

```python
yros.shape
```

```python
xros.shape
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(xros,yros,test_size=0.20,random_state=2)
```

```python
xtrain.shape,ytrain.shape,xtest.shape,ytest.shape
```

```python
from sklearn.linear_model import LogisticRegression
logr = LogisticRegression()
```

```python
logr.fit(xtrain,ytrain)
```

```python
logr.score(xtest,ytest)
```

```python
logr2 = LogisticRegression()
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(logr,xtrain,ytrain,cv=20,n_jobs=-1)
```

```python
scores.mean()
```

```python
sclist = []
for i in range( 500):
    xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=20,random_state=i)
    logr1 = LogisticRegression()
    logr1.fit(xtrain,ytrain)
    sc = logr1.score(xtest,ytest)
    sclist.append(sc)

M=max(sclist)
rs= sclist.index(M)


```
