## Non linear data


```python
import pandas as pd
df_sal = pd.read_csv(r"C:\Users\DELL\Downloads\salary_poly (2).csv")
df_sal
```

```python
df_sal = df_sal.drop("Position", axis=1)
df_sal
```

```python
import matplotlib.pyplot as plt
x = df_sal.drop("Salary",axis=1)
x
```

```python
y = df_sal["Salary"]
y
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

lr = LinearRegression()
```

```python
# data small so directly we can give x y 
lr.fit(x,y)
```

```python
lr.score(x,y)
```

```python
y_pred = lr.predict(x)
```

```python
plt.scatter(x,y,color="r")
plt.plot(x,y_pred,color="k")
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly2 = PolynomialFeatures(degree=2)
```

```python
poly2.fit(x)
```

```python
x_poly2 = poly2.transform(x)
```

```python
x_poly2
```

```python
lr1 = LinearRegression()
```

```python
lr1.fit(x_poly2,y)
```

```python
lr1.score(x_poly2,y)
```

```python
ypred2 = lr1.predict(x_poly2)
```

```python
y
```

```python
plt.scatter(x,y,color="r")
plt.plot(x,y_pred,color="y")
plt.plot(x,ypred2,color="b")
```

## degree 3

```python
poly3 = PolynomialFeatures(degree=3)
poly3.fit(x)
x_poly3=poly3.transform(x)
```

```python
lr3 = LinearRegression()
lr3.fit(x_poly3,y)
lr3.score(x_poly3,y)
```

```python
ypred3 = lr3.predict(x_poly3)
```

```python
plt.scatter(x,y,color="r")
plt.plot(x,y_pred,color="y")
plt.plot(x,ypred2,color="b")
plt.plot(x,ypred3,color="r")
```
