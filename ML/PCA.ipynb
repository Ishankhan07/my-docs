```python
col = [ 'trestbps', 'chol', 'fbs', 'thalach', 'oldpeak',  'ca', 'thal', ]
```

```python
for c in col:
    q1 = df[c].quantile(0.25)
    q3 = df[c].quantile(0.75)
    iqr = q3-q1
    lower = q1 - (1.5*iqr)
    upper = q3 + (1.5*iqr)
    df[c] = df[c].map(lambda x: lower if x<lower else upper if x>upper else x)
    # print("lower",lower)
    # print("upper",upper)
```

```python
plt.figure(figsize=(20,6))
sns.boxplot(df)
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=9)
```

```python
pca.fit(x)
```

```python
xpca= pca.transform(x)
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(xpca,y,test_size=0.20)
```
